18. UPDATE Matemático (Inflación): Debido a la inflación, los productos de la categoría "Papeles" sufrirán un aumento del 15%. Actualiza el Precio_unitario multiplicándolo por 1.15 donde la Categoría sea "Papeles".

```python
%sql
UPDATE cursodataacademy.proyecto_kiosko.data_productos
SET Precio_unitario = Precio_unitario * 1.15
WHERE lower(`Categoría`) = 'papeles'
```
