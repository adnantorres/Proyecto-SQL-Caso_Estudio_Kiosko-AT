### Sección 1: Limpieza, DDL, DML y Agregaciones Básicas (10 preguntas)
1. Modificación de Tabla (ALTER): Al cargar los datos de los tickets, la primera columna (el correlativo del 0 al 90) se cargó sin nombre o con un nombre genérico como _c0. Renombra esta columna a ID_Registro.

```python
%sql
ALTER TABLE cursodataacademy.proyecto_kiosko.data_tickets
SET TBLPROPERTIES ('delta.columnMapping.mode' = 'name');

ALTER TABLE cursodataacademy.proyecto_kiosko.data_tickets
RENAME COLUMN `Unnamed:_0` TO ID_Registro
```

2. Eliminación de Anomalías (DELETE): En la tabla de productos hay un error de sistema: el "Alfajor Chocolate" (ID 3) y la "Leche 1lt" (ID 8) aparecen duplicados. Escribe una consulta para eliminar el registro duplicado de la "Leche 1lt" asumiendo que vas a borrar temporalmente el ID 8 completo para volver a crearlo limpio.

```python
%sql
SELECT
        ID_Producto,
        count(*) AS CANTIDAD
FROM cursodataacademy.proyecto_kiosko.data_productos
GROUP BY ID_Producto
HAVING count(*) > 1
ORDER BY 2 DESC
```

```python
%sql
SELECT
        ID_Producto,
        row_number() OVER (PARTITION BY ID_Producto ORDER BY ID_Producto) AS cantidad_acum
FROM cursodataacademy.proyecto_kiosko.data_productos
ORDER BY 1 ASC
```

6. Filtros con Comodines (LIKE): El gerente quiere analizar el impacto de los chocolates. Selecciona todos los productos de tb_productos cuyo nombre contenga la palabra "Chocolate" en cualquier parte.


```python
%sql
SELECT *
FROM cursodataacademy.proyecto_kiosko.data_productos
WHERE lower(Nombre_producto) LIKE '%chocolate%'
```
