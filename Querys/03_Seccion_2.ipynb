## Sección 2: Cruce de Información - JOINs (6 preguntas) 
11. INNER JOIN (Básico): Une tb_tickets con tb_productos usando ID_Producto. Muestra el número de Ticket, la Fecha, el Nombre_producto y la Cantidad.

```python
%sql
SELECT 
    t.*,
    p.*
FROM cursodataacademy.proyecto_kiosko.data_tickets t
JOIN cursodataacademy.proyecto_kiosko.data_productos p
ON t.ID_Producto = p.ID_Producto
LIMIT 10
```

```python
%sql
SELECT 
    t.Ticket,
    t.Fecha,
    p.Nombre_producto,
    t.Cantidad
FROM cursodataacademy.proyecto_kiosko.data_tickets t
JOIN cursodataacademy.proyecto_kiosko.data_productos p
ON t.ID_Producto = p.ID_Producto
```

15. FULL OUTER JOIN: Haz un cruce completo entre productos y tickets. Esto permitirá a auditoría detectar si existen tickets que tengan un ID_Producto que ya no existe en el catálogo, o productos que no tengan tickets.

```python
%sql
SELECT
    p.Nombre_producto,
    t.Ticket,
    t.Fecha
FROM cursodataacademy.proyecto_kiosko.data_productos p
JOIN cursodataacademy.proyecto_kiosko.data_tickets t
ON p.ID_Producto = t.ID_Producto
```

```python
%sql
-- Que productos no estan dentro de la tabla de productos, pero que si se encuentran en la tabla tickets
SELECT
    p.Nombre_producto,
    t.Ticket,
    t.Fecha
FROM cursodataacademy.proyecto_kiosko.data_productos p
JOIN cursodataacademy.proyecto_kiosko.data_tickets t
ON p.ID_Producto = t.ID_Producto
WHERE p.ID_Producto IS NULL
```

```python
%sql
-- Que productos no estan dentro de la tabla de tickets, pero que si se encuentran en la tabla de productos
SELECT
    p.Nombre_producto,
    t.Ticket,
    t.Fecha
FROM cursodataacademy.proyecto_kiosko.data_productos p
JOIN cursodataacademy.proyecto_kiosko.data_tickets t
ON p.ID_Producto = t.ID_Producto
WHERE t.Ticket IS NULL
```
